# Federated SegResNet32 training: FedProx

## Local client training with FedProx proximal regularization

```python
def train_local_model(
    model,
    loader,
    loss_fn,
    device,
    lr=1e-4,
    weight_decay=1e-5,
    local_epochs=1,
    max_batches=None,
    global_reference_state=None,
    fedprox_mu=0.0,
):
    """Optimize a client model and return loss and sampling statistics."""
    model.train()

    if fedprox_mu > 0.0:
        if global_reference_state is None:
            raise ValueError("global_reference_state is required when fedprox_mu > 0")

        prox_reference = make_global_reference_state(
            global_reference_state,
            device=device,
        )
    else:
        prox_reference = None

    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=lr,
        weight_decay=weight_decay,
    )

    scaler = torch.amp.GradScaler(
        "cuda",
        enabled=use_amp,
    )

    grad_accum_steps = int(CFG.get("grad_accum_steps", 1))
    grad_clip_norm = CFG.get("grad_clip_norm", None)

    total_losses = []
    seg_losses = []
    prox_losses = []
    prox_raw_values = []
    crop_sources = []

    optimizer.zero_grad(set_to_none=True)
    accum_counter = 0

    for local_epoch in range(local_epochs):
        for step, batch in enumerate(loader):
            if max_batches is not None and step >= max_batches:
                break

            images = batch["image"].to(device, non_blocking=True)
            masks = batch["mask"].to(device, non_blocking=True)

            with torch.amp.autocast("cuda", enabled=use_amp):
                logits = model(images)
                seg_loss = loss_fn(logits, masks.unsqueeze(1))

            if prox_reference is not None:
                # Keep proximal penalty in FP32 for numerical stability.
                with torch.amp.autocast("cuda", enabled=False):
                    prox_raw = fedprox_l2_penalty(model, prox_reference)
                    prox_loss = 0.5 * float(fedprox_mu) * prox_raw
                    loss = seg_loss.float() + prox_loss
            else:
                prox_raw = torch.zeros((), device=device)
                prox_loss = torch.zeros((), device=device)
                loss = seg_loss

            loss_for_backward = loss / grad_accum_steps

            scaler.scale(loss_for_backward).backward()
            accum_counter += 1

            should_step = (accum_counter % grad_accum_steps) == 0

            if should_step:
                if grad_clip_norm is not None:
                    scaler.unscale_(optimizer)
                    torch.nn.utils.clip_grad_norm_(
                        model.parameters(),
                        max_norm=grad_clip_norm,
                    )

                scaler.step(optimizer)
                scaler.update()
                optimizer.zero_grad(set_to_none=True)

            total_losses.append(float(loss.detach().item()))
            seg_losses.append(float(seg_loss.detach().item()))
            prox_losses.append(float(prox_loss.detach().item()))
            prox_raw_values.append(float(prox_raw.detach().item()))

            if "crop_source" in batch:
                crop_sources.extend(list(batch["crop_source"]))

    # Flush remaining accumulated gradients if number of batches was not divisible
    # by grad_accum_steps.
    if accum_counter > 0 and (accum_counter % grad_accum_steps) != 0:
        if grad_clip_norm is not None:
            scaler.unscale_(optimizer)
            torch.nn.utils.clip_grad_norm_(
                model.parameters(),
                max_norm=grad_clip_norm,
            )

        scaler.step(optimizer)
        scaler.update()
        optimizer.zero_grad(set_to_none=True)

    mean_total_loss = float(np.mean(total_losses)) if len(total_losses) > 0 else float("nan")
    mean_seg_loss = float(np.mean(seg_losses)) if len(seg_losses) > 0 else float("nan")
    mean_prox_loss = float(np.mean(prox_losses)) if len(prox_losses) > 0 else float("nan")
    mean_prox_raw = float(np.mean(prox_raw_values)) if len(prox_raw_values) > 0 else float("nan")
    crop_source_counts = dict(Counter(crop_sources))

    return mean_total_loss, mean_seg_loss, mean_prox_loss, mean_prox_raw, crop_source_counts
```

## One-client FedProx smoke test

```python
test_domain = client_domains[0]
test_lr = get_round_lr(
    round_idx=1,
    total_rounds=CFG["fed_rounds"],
    base_lr=CFG["lr"],
    min_lr=CFG["min_lr"],
)

global_state_for_test = get_cpu_state_dict(global_model)

local_model = create_model().to(device)
local_model.load_state_dict(global_state_for_test, strict=True)

(
    test_total_loss,
    test_seg_loss,
    test_prox_loss,
    test_prox_raw,
    test_crop_counts,
) = train_local_model(
    model=local_model,
    loader=client_loaders[test_domain],
    loss_fn=loss_fn,
    device=device,
    lr=test_lr,
    weight_decay=CFG["weight_decay"],
    local_epochs=1,
    max_batches=2,
    global_reference_state=global_state_for_test,
    fedprox_mu=CFG["fedprox_mu"],
)

del local_model

if torch.cuda.is_available():
    torch.cuda.empty_cache()

print("test domain:", test_domain)
print("test lr:", test_lr)
print("fedprox_mu:", CFG["fedprox_mu"])
print("test total loss:", test_total_loss)
print("test seg loss:", test_seg_loss)
print("test prox loss:", test_prox_loss)
print("test prox raw L2:", test_prox_raw)
print("test crop counts:", test_crop_counts)
```

## FedProx training loop

```python
history = []
best_val_loss = float('inf')
best_mean_dice = -float('inf')
best_round = None
rounds_without_improvement = 0
for round_idx in range(1, CFG['fed_rounds'] + 1):
    print(f"\n========== FedProx round {round_idx}/{CFG['fed_rounds']} ==========")
    round_lr = get_round_lr(
        round_idx=round_idx,
        total_rounds=CFG['fed_rounds'],
        base_lr=CFG['lr'],
        min_lr=CFG['min_lr']
    )
    print(f"round_lr={round_lr:.2e} | fedprox_mu={CFG['fedprox_mu']:.2e}")
    global_state = get_cpu_state_dict(global_model)
    client_states = {}
    client_total_losses = {}
    client_seg_losses = {}
    client_prox_losses = {}
    client_prox_raw_values = {}
    client_crop_counts = {}
    for domain in client_domains:
        print(f'\nClient: {domain}')
        local_model = create_model().to(device)
        local_model.load_state_dict(global_state, strict=True)
        local_total_loss, local_seg_loss, local_prox_loss, local_prox_raw, crop_counts = train_local_model(
            model=local_model,
            loader=client_loaders[domain],
            loss_fn=loss_fn,
            device=device,
            lr=round_lr,
            weight_decay=CFG['weight_decay'],
            local_epochs=CFG['local_epochs'],
            max_batches=CFG['max_client_batches'],
            global_reference_state=global_state,
            fedprox_mu=CFG['fedprox_mu']
        )
        client_total_losses[domain] = local_total_loss
        client_seg_losses[domain] = local_seg_loss
        client_prox_losses[domain] = local_prox_loss
        client_prox_raw_values[domain] = local_prox_raw
        client_crop_counts[domain] = crop_counts
        client_states[domain] = get_cpu_state_dict(local_model)
        del local_model
        if torch.cuda.is_available():
            torch.cuda.empty_cache()
        print(f'{domain} total_loss={local_total_loss:.4f} | seg_loss={local_seg_loss:.4f} | prox_loss={local_prox_loss:.6f} | prox_raw={local_prox_raw:.2f} | crops={crop_counts}')
    avg_state = weighted_average_state_dicts(client_states=client_states, client_weights=client_weights)
    global_model.load_state_dict(avg_state, strict=True)
    global_model.to(device)
    val_loss, val_dice = validate_patch_level(
        model=global_model,
        loader=val_loader,
        loss_fn=loss_fn,
        device=device,
        max_batches=CFG['max_val_batches'],
        use_amp=use_amp
    )
    mean_dice = float(np.mean([val_dice['WT'], val_dice['TC'], val_dice['ET']]))
    row = {
        'round': round_idx,
        'lr': round_lr,
        'fedprox_mu': CFG['fedprox_mu'],
        'val_loss': val_loss,
        'val_dice_WT': val_dice['WT'],
        'val_dice_TC': val_dice['TC'],
        'val_dice_ET': val_dice['ET'],
        'val_mean_dice': mean_dice
    }
    for domain in client_domains:
        row[f'local_loss_{domain}'] = client_total_losses[domain]
        row[f'local_seg_loss_{domain}'] = client_seg_losses[domain]
        row[f'local_prox_loss_{domain}'] = client_prox_losses[domain]
        row[f'local_prox_raw_{domain}'] = client_prox_raw_values[domain]
        row[f'crop_source_counts_{domain}'] = json.dumps(client_crop_counts[domain])
    history.append(row)
    improved_mean_dice = mean_dice > best_mean_dice + CFG['early_stop_min_delta']
    if improved_mean_dice:
        best_mean_dice = mean_dice
        best_round = round_idx
        rounds_without_improvement = 0
    else:
        rounds_without_improvement += 1
    print(f"\nRound {round_idx:03d} | val_loss={val_loss:.4f} | mean_dice={mean_dice:.4f} | WT={val_dice['WT']:.4f} | TC={val_dice['TC']:.4f} | ET={val_dice['ET']:.4f} | no_improve={rounds_without_improvement}")
    ckpt_last_path = os.path.join(CFG['save_dir'], 'fedprox_last.pt')
    torch.save(
        {
            'model': global_model.state_dict(),
            'round': round_idx,
            'cfg': CFG,
            'history': history,
            'client_weights': client_weights,
            'client_sizes': client_sizes,
            'best_round': best_round,
            'best_mean_dice': best_mean_dice,
            'algorithm': 'FedProx',
            'fedprox_mu': CFG['fedprox_mu']
        },
        ckpt_last_path
    )
    if val_loss < best_val_loss:
        best_val_loss = val_loss
        ckpt_best_loss_path = os.path.join(CFG['save_dir'], 'fedprox_best_loss.pt')
        torch.save(
            {
                'model': global_model.state_dict(),
                'round': round_idx,
                'cfg': CFG,
                'history': history,
                'client_weights': client_weights,
                'client_sizes': client_sizes,
                'best_val_loss': best_val_loss,
                'best_round': best_round,
                'best_mean_dice': best_mean_dice,
                'algorithm': 'FedProx',
                'fedprox_mu': CFG['fedprox_mu']
            },
            ckpt_best_loss_path
        )
        print('saved best loss:', ckpt_best_loss_path)
    if improved_mean_dice:
        ckpt_best_dice_path = os.path.join(CFG['save_dir'], 'fedprox_best_mean_dice.pt')
        torch.save(
            {
                'model': global_model.state_dict(),
                'round': round_idx,
                'cfg': CFG,
                'history': history,
                'client_weights': client_weights,
                'client_sizes': client_sizes,
                'best_mean_dice': best_mean_dice,
                'best_round': best_round,
                'algorithm': 'FedProx',
                'fedprox_mu': CFG['fedprox_mu']
            },
            ckpt_best_dice_path
        )
        print('saved best mean dice:', ckpt_best_dice_path)
    if round_idx >= CFG['min_rounds'] and rounds_without_improvement >= CFG['early_stop_patience']:
        print(f'Early stopping at round {round_idx}. Best round: {best_round}, best_mean_dice={best_mean_dice:.4f}')
        break
```
